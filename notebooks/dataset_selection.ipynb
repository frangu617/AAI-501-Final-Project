## 3. Online Shoppers Purchasing Intention — recommended

**Proposed question:** Can browsing behavior predict whether a session ends in a purchase, and can a smaller set of features retain useful predictive performance?

**Application:** Support analysis of conversion patterns and visitor segments for an e-commerce business. The dataset does not establish that any marketing intervention causes purchases.

**Why use it:** Its 12,330 sessions are large enough for the assignment and manageable on a laptop. Its 17 predictors allow feature interpretation without an excessive feature-engineering workload. UCI lists both classification and clustering as associated tasks. The target has 1,908 purchasing sessions and 10,422 non-purchasing sessions, so class imbalance creates a meaningful evaluation issue.

**Proposed experiments:**

1. Compare a majority-class baseline, logistic regression, and random forest using the same stratified training/validation/test partitions.
2. Tune a small number of parameters using cross-validation on training data only.
3. Compare full-feature models with reduced-feature models selected within training folds.
4. Optionally cluster scaled browsing features with K-means, excluding the target, then describe conversion rates across clusters. Compare several cluster counts using silhouette scores and interpretability.

**Evaluation:** Report purchase-class precision, recall, F1, PR-AUC (state whether using average precision), ROC-AUC, and a confusion matrix. Accuracy alone is insufficient because approximately 84.5% of sessions do not end in purchases. Choose any classification threshold using validation data.

**Preprocessing and limitations:** Encode nominal variables such as month, visitor type, browser, and traffic type appropriately; numeric category codes should not automatically imply order. Scale numeric inputs for logistic regression and K-means. Check skew and duplicates. Define the prediction time: these are session summaries, not an event stream for demonstrating early-session forecasting. Audit `PageValues` and other features for availability at prediction time; include a sensitivity experiment without `PageValues` if its construction cannot be verified. Fit all transformations on training data only.

**Possible deliverable:** A notebook explaining conversion drivers, classifier comparisons, a reduced-feature experiment, and interpretable visitor segments.

[Official UCI documentation and download](https://archive.ics.uci.edu/dataset/468/online+shoppers+purchasing+intention+dataset)

## 4. Bank Marketing — strongest campaign-targeting story

**Proposed question:** Can information available before a marketing call predict term-deposit subscriptions, and which features provide the most useful ranking of potential subscribers?

**Application:** Help a bank prioritize campaign contacts under a limited contact budget.

**Why use it:** The full legacy file has 45,211 records and a clear binary outcome. Mixed customer and campaign variables support interpretable baselines, nonlinear models, feature selection, and optional customer segmentation.

**Choose one version:** Use `bank-full.csv` consistently: 45,211 rows and 16 predictors plus `y`. The separate `bank-additional-full.csv` has 41,188 rows and 20 predictors. Do not mix their schemas or report one version's dimensions for the other. The files use semicolon delimiters.

**Proposed experiments:**

1. Compare logistic regression and a decision tree or random forest against a majority-class baseline.
2. Compare all eligible predictors with a smaller feature set selected using training data.
3. Evaluate subscription recall or lift among the highest-scoring 10% or 20% of contacts, alongside conventional classification metrics.
4. Optionally cluster pre-call customer characteristics and describe subscription rates by segment without including `y` in clustering.

**Evaluation:** PR-AUC, precision, recall, F1, ROC-AUC, confusion matrices, and lift at a fixed contact budget. Use a held-out later block if modeling future campaign performance from the date-ordered file; use validation that respects ordering where feasible. Do not infer financial ROI without justified cost and revenue assumptions.

**Preprocessing and limitations:** Exclude `duration` for pre-call prediction because call length is only known afterward. Audit current-campaign fields such as `campaign` against the exact prediction point. Treat `unknown` categories explicitly. In the legacy file, `pdays = -1` means the client was not previously contacted; represent that meaning deliberately. Encode categories and scale numeric inputs where required. Historical associations do not establish campaign effectiveness or present-day generalization.

**Possible deliverable:** A campaign-prioritization prototype with interpretable drivers and an analysis of predictive quality under a fixed contact budget.

[Official UCI documentation and download](https://archive.ics.uci.edu/dataset/222/bank+marketing)

## 5. Bike Sharing — strongest regression option

**Proposed question:** How accurately can calendar and weather information predict hourly rental demand, and when do prediction errors become largest?

**Application:** Support system-wide fleet availability and staffing planning. The dataset lacks station-level information, so it cannot directly support station-specific rebalancing recommendations.

**Why use it:** Hourly observations comfortably exceed 1,000 examples, the target is easy to explain, and seasonal and hourly patterns support useful visualizations. Use `hour.csv`; `day.csv` has only 731 observations and fails the project's minimum-size rule by itself.

**Proposed experiments:**

1. Compare an hour-of-week historical-average baseline, linear or ridge regression, and random forest regression.
2. Compare calendar-only predictors against calendar-plus-weather predictors.
3. Compare full and reduced feature sets, and examine errors during peak hours, weekends, and weather conditions.
4. Optionally cluster daily demand profiles to identify recurring patterns; make clear that these are descriptive profiles, not predictors available before a day unfolds.

**Evaluation:** MAE and RMSE on a final chronological holdout, with rolling or expanding-window validation for tuning. Plot actual versus predicted demand, residuals over time, and error by hour. If adding peak-demand classification, define the peak threshold from training data only.

**Preprocessing and limitations:** Parse dates, drop the row identifier `instant`, and encode calendar categories appropriately. Exclude `casual` and `registered` when predicting `cnt`: they sum exactly to the target. Fit transformations on training periods. Actual recorded weather supports demand estimation under observed conditions; a future operational forecast would require weather forecasts available at prediction time. Data from 2011–2012 may not represent current demand.

**Possible deliverable:** A demand-prediction notebook with chronological evaluation, peak-period error analysis, and an explanation of the value added by weather information.

[Official UCI documentation and download](https://archive.ics.uci.edu/dataset/275/bike+sharing+dataset)
